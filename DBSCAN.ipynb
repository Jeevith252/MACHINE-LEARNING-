```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons # to create datasets
from sklearn.cluster import DBSCAN , KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X , y_true = make_moons(n_samples= 500 , noise = 0.05 , random_state= 42) # noise is the distance between each and every point
```

```python
df = pd.DataFrame ( X , columns = ['Feature_1' , 'Feature_2'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters = 3 , random_state= 42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x = df['Feature_1'] , y =df['Feature_2'] ,hue = df['kmeans_cluster'] , palette= 'tab10')
```
